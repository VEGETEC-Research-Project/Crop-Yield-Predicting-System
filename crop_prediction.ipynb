```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Load dataset
df = pd.read_csv("crop_dataset.csv")

# Encode categorical variable (Crop_Type: Tomato -> 0, Capsicum -> 1)
encoder = LabelEncoder()
df["Crop_Type"] = encoder.fit_transform(df["Crop_Type"])

# Select features and target
X = df.drop(columns=["Crop_Type"])  # Features
y = df["Crop_Type"]  # Target variable

# Feature Scaling (Standardization)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split dataset into training (80%) and testing (20%) sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

print("Preprocessing Completed!")
print(f"Training Set: {X_train.shape}, Testing Set: {X_test.shape}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Initialize Logistic Regression model
model = LogisticRegression()

# Train the model
model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)

# Evaluate model performance
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
report = classification_report(y_test, y_pred)

print(f"Model Accuracy: {accuracy:.2f}")
print("\nConfusion Matrix:\n", conf_matrix)
print("\nClassification Report:\n", report)
```

```python
import pickle

# Save the trained model
model_filename = "crop_prediction_model.pkl"
with open(model_filename, "wb") as file:
    pickle.dump(model, file)

print(f"Model saved as {model_filename}")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
import pandas as pd

# Load two datasets
df_tomato = pd.read_csv("tomato.csv")
df_capsicum = pd.read_csv("capsicum.csv")

# Combine both datasets into one
df = pd.concat([df_tomato, df_capsicum], ignore_index=True)

# Encode categorical variable (Crop_Type: Tomato -> 0, Capsicum -> 1)
encoder = LabelEncoder()
df["Crop_Type"] = encoder.fit_transform(df["Crop_Type"])

# Select features and target
X = df.drop(columns=["Crop_Type","Date","Soil_Type","Wind_Speed","Soil_Quality","Area","Crop_Yield"])  # Features
y = df["Crop_Type"]  # Target variable

# Feature Scaling (Standardization)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split dataset into training (80%) and testing (20%) sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

print("Preprocessing Completed!")
print(f"Training Set: {X_train.shape}, Testing Set: {X_test.shape}")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Initialize Logistic Regression model
model = LogisticRegression()

# Train the model
model.fit(X_train, y_train)

# Predictions
y_pred = model.predict(X_test)

# Evaluate model performance
accuracy = accuracy_score(y_test, y_pred)
conf_matrix = confusion_matrix(y_test, y_pred)
report = classification_report(y_test, y_pred)

print(f"Model Accuracy: {accuracy:.2f}")
print("\nConfusion Matrix:\n", conf_matrix)
print("\nClassification Report:\n", report)
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
import joblib

# Load dataset
capsicum_df = pd.read_csv("capsicum.csv")
tomato_df = pd.read_csv("tomato.csv")

# Add a Crop column
tomato_df["Crop"] = "Tomato"
capsicum_df["Crop"] = "Capsicum"

# Select features
features = ["Temperature", "Humidity", "Soil_pH", "N", "P", "K", "Crop"]
crop_data = pd.concat([tomato_df[features], capsicum_df[features]], ignore_index=True)

# Shuffle dataset
crop_data = crop_data.sample(frac=1, random_state=42).reset_index(drop=True)

# Prepare features and labels
X = crop_data.drop(columns=["Crop"])
y = crop_data["Crop"]

# Encode labels
label_encoder = LabelEncoder()
y_encoded = label_encoder.fit_transform(y)

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y_encoded, test_size=0.2, random_state=42)

# Standardize features
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train the model
model = RandomForestClassifier(n_estimators=200, random_state=42, max_depth=10)
model.fit(X_train_scaled, y_train)

# Make predictions
y_pred = model.predict(X_test_scaled)

# Evaluate the model
accuracy = accuracy_score(y_test, y_pred)
print(f"Model Accuracy: {accuracy:.2%}")
print(classification_report(y_test, y_pred, target_names=label_encoder.classes_))

# Save the model and encoders
joblib.dump(model, "crop_prediction_model.pkl")
joblib.dump(scaler, "scaler.pkl")
joblib.dump(label_encoder, "label_encoder.pkl")

print("Model and encoders saved successfully!")

# Load the model for testing
def predict_crop(temp, humidity, ph, n, p, k):
    model = joblib.load("crop_prediction_model.pkl")
    scaler = joblib.load("scaler.pkl")
    label_encoder = joblib.load("label_encoder.pkl")
    
    input_data = np.array([[temp, humidity, ph, n, p, k]])
    input_scaled = scaler.transform(input_data)
    prediction = model.predict(input_scaled)
    predicted_crop = label_encoder.inverse_transform(prediction)[0]
    
    return predicted_crop

# Example test case
example_prediction = predict_crop(23.83013916, 75.28104911,5.906320077, 184.1614516, 100.0255129, 57.25166425)
print(f"Predicted Crop: {example_prediction}")
```
